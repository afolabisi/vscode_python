```python
import pandas as pd
import matplotlib.pyplot
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df= dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_usa = df[df['job_country']== 'United States'].copy()

df_usa
```

```python
df_usa['job_post_month'] = df['job_posted_date'].dt.strftime('%B')
df
```

```python
df_us_pivot = df_usa.pivot_table(index = 'job_post_month', columns = 'job_title_short', aggfunc= 'size')
```

```python
df_us_pivot.reset_index(inplace= True)
df_us_pivot['month_no'] =pd.to_datetime(df_us_pivot['job_post_month'],format='%B').dt.month
df_us_pivot.sort_values('month_no', inplace=True)
df_us_pivot.set_index('job_post_month', inplace= True)
df_us_pivot.drop(columns='month_no', inplace=True)
```

```python
df_us_pivot
```

```python
df_us_pivot.plot(kind='line')
```

```python
top3=df_usa['job_title_short'].value_counts().head(3)
top_3=top3. index.to_list()
```

```python
df_us_pivot[top_3].plot(kind= 'line')
```
